```python
from ucimlrepo import fetch_ucirepo

lenses = fetch_ucirepo(id=58)

x = lenses.data.features
y = lenses.data.targets
```

```python
x
```

```python
y
```

```python
print("Formato de X: ", x.shape)
print("Formato de y: ", y.shape)

print("\nColunas de X:")
print(x.columns)

print("\nClasses:")
print(y.iloc[:, 0].unique())
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(handle_unknown='ignore')

X_encoded = encoder.fit_transform(x)

print(X_encoded)
```

```python
print("Formato original:", x.shape)
print("Formato após One-Hot Encoding:", X_encoded.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
modelo = LogisticRegression(max_iter=1000)
```

```python
modelo.fit(X_encoded, y.values.ravel())
```

```python
y_pred = modelo.predict(X_encoded)
```

```python
print("Valores reais")
print(y.values.ravel())

print("\nValores previstos: ")
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score

acuracia = accuracy_score (y.values.ravel(), y_pred)

print(f"Percentual de acertos: {acuracia * 100:.2f}%")
```

```python
print("======================================")
print("   CLASSIFICAÇÃO DO DATASET LENSES")
print("======================================")
print(f"Quantidade de amostras: {len(x)}")
print(f"Quantidade de características: {x.shape[1]}")
print(f"Quantidade de classes: {len(y.iloc[:, 0].unique())}")
print(f"Percentual de acertos: {acuracia * 100:.2f}%")
print("======================================")
```

**Análise Final dos Resultados**

Neste exercício foi utilizado o algoritmo de Regressão Logística, por meio da biblioteca Scikit-Learn, para realizar a classificação do conjunto de dados Lenses. O conjunto utilizado possui 24 amostras, três características de entrada (age, spectacle_prescription e astigmatic) e três classes de saída.

Como as características utilizadas são categóricas, foi aplicado o método One-Hot Encoding, permitindo que os dados fossem utilizados adequadamente pelo modelo de classificação.

Após o treinamento, foram realizadas previsões sobre o próprio conjunto de dados utilizado no treinamento. O modelo classificou corretamente 22 das 24 amostras, resultando em uma acurácia de 91,67%.

Esse resultado indica que o modelo apresentou um bom desempenho na classificação das amostras utilizadas no treinamento, embora tenha ocorrido erro em duas previsões. É importante destacar que essa avaliação foi realizada sobre os próprios dados utilizados para treinar o modelo, conforme solicitado no exercício, portanto o resultado não representa necessariamente o desempenho do modelo em dados novos.
