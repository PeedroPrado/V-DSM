```python
from sklearn.datasets import fetch_openml
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import precision_score
from sklearn.metrics import accuracy_score
from sklearn.metrics import recall_score
```

```python
mnist = fetch_openml('mnist_784', version = 1, as_frame = False)
```

```python
x = mnist["data"]
y = mnist ["target"]
```

```python
print(x.shape)
print(y.shape)
```

```python
import matplotlib.pyplot as plt
```

```python
imagem = x[0]

plt.imshow(imagem.reshape(28, 28), cmap="binary")
plt.axis("off")
plt.show()
```

```python
print(y[0])
```

```python
imagem = x[10]

plt.imshow(imagem.reshape(28, 28), cmap="binary")
plt.axis("off")
plt.show()
```

```python
print(y[10])
```

```python
x_train = x[:6000]
x_test = x[6000:]

y_train = y[:6000]
y_test = y[6000:]

```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
y_train_3 = (y_train == '3')
y_test_3 = (y_test == '3')
```

```python
print(y_train[:20])
print(y_train_3[:20])
```

```python
sgd_clf = SGDClassifier(random_state= 42)
```

```python
sgd_clf.fit(x_train, y_train_3)
```

```python
imagem_teste = x_test[0]

plt.imshow(imagem_teste.reshape(28, 28), cmap="binary")
plt.axis("off")
plt.show()
```

```python
sgd_clf.predict([imagem_teste])
```

```python
print(y_test[0])
```

```python
y_pred = sgd_clf.predict(x_test)
```

```python
matriz = confusion_matrix(y_test_3, y_pred)
print(matriz)
```

```python
precisao = precision_score(y_test_3, y_pred)
print("Precisão: ", precisao)
```

```python
acuracia = accuracy_score(y_test_3, y_pred)

print("Acurácia:", acuracia)
```

```python
sensibilidade = recall_score(y_test_3, y_pred)

print("Sensibilidade:", sensibilidade)
```

```python
print(f"Precisão: {precisao:.2%}")
print(f"Acurácia: {acuracia:.2%}")
print(f"Sensibilidade (Recall): {sensibilidade:.2%}")
```
